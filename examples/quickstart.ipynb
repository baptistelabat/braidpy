## A first braid

A braid is a word in Artin's generators: `1` is σ₁, `-1` its inverse, and
so on.

```python
from braidpy import Braid

b = Braid([1, 2, -1])

b.format()
```

```python
b.format_to_notation(target="alpha")
```

```python
b.draw()
```

```python
b.plot()
```

## Is it the same braid?

The same braid can be written many ways. braidpy compares braids by their
topology:

```python
from braidpy import tools

tools.compare_braids("1 -2", "-2 1")
```

## How to make it

And it says how to make a braid: on a kumihimo disk, a Jacquard lace
machine's punchcard, or Ashley's stand:

```python
ways = tools.ways_to_make("1 -2")

ways["disk"]["moves"]
```
